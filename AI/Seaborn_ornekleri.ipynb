```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
print(sns.get_dataset_names()) #seaborn kütüphanesindeki datasetlerin isimlerini verir
```

```python
bahsis_veri = sns.load_dataset('tips') #tips datasetini çekmek için kullanılır.
print(bahsis_veri)
```

```python
bahsis_veri.rename(columns={'total_bill' : 'fatura_toplami',
                           'tip'         : 'bahsis',
                           'sex'         : 'cinsiyet',
                           'smoker'      : 'sigara',
                           'day'         : 'gun',
                           'time'        : 'zaman',
                           'size'        : 'kisi_sayisi'}, inplace = True)
# bu komut ile datasetin sütun isimleri değiştirilir, değişikliğin datasette kalıcı olmasını sağlamak için inplace
# parametresi True olmalıdır.
```

```python
print(bahsis_veri)
```

```python
print(bahsis_veri.info()) #dataseti ile ilgili bilgileri verir
```

```python
print(bahsis_veri.describe()) #istatistik verileri görüntüler
```

Veri Görselleştirme

```python
sns.scatterplot(data = bahsis_veri, x='fatura_toplami' , y = 'bahsis')
```

```python
sns.set_style("dark")  #darkgrid , whitegrid, dark, white, ticks
sns.set(rc={'figure.figsize':(8,6)}) #grafiğin boyutunu ayarlar
sns.scatterplot(data = bahsis_veri, x='fatura_toplami' , y = 'bahsis' , hue='cinsiyet' , color = 'b' , palette = 'Set1')
plt.ylabel("Bahşiş") #Y ekseni başlığı
plt.xlabel("Fatura Toplamı") #X ekseni başlığı
plt.title("Cinsiyete Göre Fatura Toplamı ve Bahşiş Saçılımı") #Grafik başlığı
plt.show() #Görüntüle
```

```python
sns.relplot(data = bahsis_veri, x='fatura_toplami' , y = 'bahsis' , hue = 'cinsiyet' , 
            style = 'zaman',  palette='Set2' , height=6)
plt.ylabel("Bahşiş") #Y ekseni başlığı
plt.xlabel("Fatura Toplamı") #X ekseni başlığı
plt.title("Cinsiyete Göre Fatura Toplamı ve Bahşiş Saçılımı (Cinsiyet - Zaman)") #Grafik başlığı
plt.show() #Görüntüle
```

```python
sns.set(rc={'figure.figsize':(8,6)})
sns.relplot(x = 'fatura_toplami' , y='bahsis', size='kisi_sayisi', sizes=(5,200), 
            hue='kisi_sayisi' , height=6, data= bahsis_veri)
plt.ylabel("Bahşiş") #Y ekseni başlığı
plt.xlabel("Fatura Toplamı") #X ekseni başlığı
plt.title("Cinsiyete Göre Fatura Toplamı ve Bahşiş Saçılımı (Cinsiyet - Zaman)") #Grafik başlığı
plt.show() #Görüntüle
```
