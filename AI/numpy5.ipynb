```python
import numpy as np
```

```python
x = np.random.randint(10,30,10)
print(x)
```

```python
y = np.sort(x)
print(y)
```

```python
x2 = np.random.randint(10,20,(4,3))
print(x2)
```

```python
y2 = np.sort(x2) #satır bazlı sıralama yapar
print(y2)
```

```python
y3 = np.sort(x2,axis=0)  #sütun bazlı sıralama yapar
print(y3)
```

```python
a = np.arange(1,11)
print(a)
```

```python
print(a+5)
print(a-3)
print(a*3)
print(a/2)
print(a**2)
```

```python
q = np.array([10,20,30])
p = np.array([3,4,5])
print(np.add(q,p))
```

```python
print(np.subtract(q,p))
```

```python
print(np.multiply(q,p))
```

```python
print(np.power(q,p))
```

```python
dizi1 = np.random.randint(1,100,(4,3))
print(dizi1)
```

```python
print(dizi1.max())
print(dizi1.max(axis=0))
print(dizi1.max(axis=1))
```

```python
print(dizi1.min())
print(dizi1.min(axis=0))
print(dizi1.min(axis=1))
```

```python
print(dizi1.sum())
print(dizi1.sum(axis=0))
print(dizi1.sum(axis=1))
```

```python
print(dizi1.mean())
print(dizi1.mean(axis=0))
print(dizi1.mean(axis=1))
```
